```python
import numpy as np
a=np.array([10,20,30,40,50])
print("Array: ",a)
print("Addition: ",a+5)
print("Subtraction: ", a-5)
print("Multiplication: ",a*5)
print("Division: ",a/5)
print("Modulus: ",a%5)
print("Exponentiation: ",a**5)
```

```python
# Array indexing
import numpy as np
a=np.array([10,20,30,40,50])
print("First element: ",a[0])
print("Third element: ",a[2])
print("Last element: ",a[-1])
```

```python
# Array slicing
import numpy as np
a=np.array([10,20,30,40,50])
print("First three elements: ",a[0:3])
print("Elements from index 2: ", a[2:])
```

```python
import numpy as np
marks=([65,70,75,80,85])
print("Marks: ",marks)
print("Sum:",np.sum(marks))
print("Mean:",np.mean(marks))
print("Maximum:",np.max(marks))
print("Minimum:",np.min(marks))
```

```python
# Complete Simple Program
import numpy as np
marks=np.array([65,70,75,80,85])
print("Marks: ",marks)
print("No of elements: ",marks.size)
print("Dimension: ",marks.ndim)
print("Marks + 5:",marks+5)
print("Marks * 5:",marks*5)
print("Total Marks:",np.sum(marks))
print("Average Marks:",np.mean(marks))
print("Maximum:",np.max(marks))
print("Minimum:",np.min(marks))
```
